# Analysing the BioSystem execution with Hybrid functionals
With this notebook we would like to extract the requirements for running the 
PBE0 hybrid functional with all the related input files of the `input` directory.
We will use the SystemCalculator in dry-run mode to do so.

```python
from futile.Utils import file_list
inputs = file_list(directory='input',suffix='yaml',include_directory_path=True)
from BigDFT.Calculators import SystemCalculator
import yaml
from BigDFT.Inputfiles import Inputfile
code = SystemCalculator(dry_run=True)
logs={}
for inp in inputs:
    with open(inp) as ifile:
        dinp = yaml.safe_load(ifile)
    I = Inputfile(dinp)
    I.set_psp_krack('PBE')
    I.set_psp_nlcc()
    logs[inp] = code.run(input=I)
```

```python
from numpy import array
estimated={}
for name,log in logs.items():
    d=array(log.log['Sizes of the simulation domain']['Grid Spacing Units'])
    grid = 2*d+31
    orb = log.log['Total Number of Orbitals']
    estimated[log.memory_peak] = dict(name=name,orbitals=orb,grid=grid)
    
```

```python
from os.path import basename
for mem in sorted(estimated):
    cl=estimated[mem]
    print(basename(cl['name']), ' -- estimated peak for BigDFT:', mem)
    print('-n ',cl['grid'],' -o [',cl['orbitals'],'] -g F')
```
